## Identify and Handle Missing Values

### Subtask:
Check for missing values in the dataset, especially in 'contact', 'pdays', and 'poutcome' as indicated by the metadata, and decide on an appropriate strategy for handling them (e.g., imputation, removal, or marking as 'unknown').


**Reasoning**:
To begin identifying and handling missing values, I will calculate and display the number and percentage of missing values for each column in the features DataFrame X. This will provide a clear overview of which columns have missing data and to what extent, guiding the subsequent handling strategy.



```
missing_values_count = X.isnull().sum()
missing_values_percentage = (X.isnull().sum() / len(X)) * 100

missing_info_X = pd.DataFrame({
    'Missing Count': missing_values_count,
    'Missing Percentage': missing_values_percentage
})

print("Missing values in features (X) dataset:")
print(missing_info_X[missing_info_X['Missing Count'] > 0])
```

**Reasoning**:
As instructed, I will now replace missing values (NaN) in the specified categorical columns ('job', 'education', 'contact', 'poutcome') with the string 'unknown'. This strategy is appropriate for categorical data where missingness might represent an unrecorded category rather than an absence of information that could be imputed numerically.



```
X['job'] = X['job'].fillna('unknown')
X['education'] = X['education'].fillna('unknown')
X['contact'] = X['contact'].fillna('unknown')
X['poutcome'] = X['poutcome'].fillna('unknown')

print("Missing values after filling with 'unknown':")
print(X.isnull().sum()[X.isnull().sum() > 0])
```

**Reasoning**:
The previous code produced a `SettingWithCopyWarning`. To explicitly ensure modifications are made on a copy of the DataFrame and prevent this warning, I will explicitly create a copy of `X` before performing the `fillna` operations. This ensures that changes are made to a standalone DataFrame, and completes the missing value imputation for categorical columns as specified in the subtask.



```
X = X.copy() # Explicitly create a copy to avoid SettingWithCopyWarning
X['job'] = X['job'].fillna('unknown')
X['education'] = X['education'].fillna('unknown')
X['contact'] = X['contact'].fillna('unknown')
X['poutcome'] = X['poutcome'].fillna('unknown')

print("Missing values after filling with 'unknown':")
print(X.isnull().sum()[X.isnull().sum() > 0])
```

## Visualize Key Feature Relationships

### Subtask:
Generate graphical presentations to visualize the distributions of key features and their relationships with the target variable 'y'. This will include bar plots for categorical variables and histograms/box plots for numerical variables, ensuring all plots have legends for clarity.


**Reasoning**:
To begin visualizing key feature relationships, I will first identify all categorical columns in the `X` DataFrame to prepare for generating bar plots showing their distribution against the target variable 'y'.



```
categorical_cols = X.select_dtypes(include='object').columns

print("Categorical columns identified:")
print(categorical_cols.tolist())
```

**Reasoning**:
Now that categorical columns have been identified, I will identify all numerical columns in the `X` DataFrame. This completes the first instruction of the subtask and prepares for generating appropriate plots for both types of features.



```
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns

print("Numerical columns identified:")
print(numerical_cols.tolist())
```

**Reasoning**:
Now that both categorical and numerical columns have been identified, I will proceed with generating bar plots for each categorical feature against the target variable 'y'. This will help visualize their distributions and relationships with the subscription outcome, addressing the second instruction of the subtask.



```
import matplotlib.pyplot as plt
import seaborn as sns

# Combine X and y for easier plotting
df_combined = pd.concat([X, y], axis=1)

print("Generating bar plots for categorical features...")

for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(x=col, hue='y', data=df_combined, palette='viridis')
    plt.title(f'Distribution of {col} by Target Variable (y)')
    plt.xlabel(col.replace('_', ' ').capitalize())
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
    plt.legend(title='Subscribed', loc='upper right')
    plt.tight_layout()
    plt.show()

print("Bar plots for categorical features generated.")
```

**Reasoning**:
Now that bar plots for categorical features have been generated, I will generate a histogram and a box plot for each numerical feature. The histogram will show the overall distribution, and the box plot will visualize its distribution across the 'yes' and 'no' categories of the target variable 'y', completing the visualization part of the subtask.



```
print("Generating histograms and box plots for numerical features...")

for col in numerical_cols:
    # Create a figure with two subplots side-by-side
    fig, axes = plt.subplots(1, 2, figsize=(18, 6))

    # Histogram for overall distribution
    sns.histplot(x=col, data=df_combined, kde=True, ax=axes[0], palette='viridis')
    axes[0].set_title(f'Distribution of {col}')
    axes[0].set_xlabel(col.replace('_', ' ').capitalize())
    axes[0].set_ylabel('Frequency')

    # Box plot for distribution across target variable 'y'
    sns.boxplot(x='y', y=col, data=df_combined, ax=axes[1], palette='viridis')
    axes[1].set_title(f'{col} Distribution by Target Variable (y)')
    axes[1].set_xlabel('Subscription Status')
    axes[1].set_ylabel(col.replace('_', ' ').capitalize())

    plt.tight_layout()
    plt.show()

print("Histograms and box plots for numerical features generated.")
```

**Reasoning**:
The previous cell generated `UserWarning` for `sns.histplot` and `FutureWarning` for `sns.boxplot` due to incorrect `palette` usage when `hue` was not explicitly assigned. To fix this, I will remove `palette` from `sns.histplot` as no `hue` is used, and for `sns.boxplot`, I will explicitly assign `hue='y'` and `legend=False` as suggested by the warning to correctly apply the palette to the target variable categories.



```
print("Generating histograms and box plots for numerical features...")

for col in numerical_cols:
    # Create a figure with two subplots side-by-side
    fig, axes = plt.subplots(1, 2, figsize=(18, 6))

    # Histogram for overall distribution
    sns.histplot(x=col, data=df_combined, kde=True, ax=axes[0]) # Removed palette as no hue is used
    axes[0].set_title(f'Distribution of {col}')
    axes[0].set_xlabel(col.replace('_', ' ').capitalize())
    axes[0].set_ylabel('Frequency')

    # Box plot for distribution across target variable 'y'
    sns.boxplot(x='y', y=col, data=df_combined, ax=axes[1], palette='viridis', hue='y', legend=False) # Added hue='y' and legend=False
    axes[1].set_title(f'{col} Distribution by Target Variable (y)')
    axes[1].set_xlabel('Subscription Status')
    axes[1].set_ylabel(col.replace('_', ' ').capitalize())

    plt.tight_layout()
    plt.show()

print("Histograms and box plots for numerical features generated.")
```
