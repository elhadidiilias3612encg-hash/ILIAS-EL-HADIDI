This dataset is related to direct marketing campaigns (phone calls) of a Portuguese banking institution.

### **Dataset Overview:**
*   **Name:** Bank Marketing
*   **Abstract:** The primary goal is to predict if a client will subscribe to a term deposit (variable 'y').
*   **Number of Instances:** 45,211
*   **Number of Features:** 16
*   **Tasks:** Classification
*   **Missing Values:** Yes, indicated by 'NaN' (for features like `contact`, `pdays`, `poutcome`)

### **Key Variables Information:**
*   **`age`**: Numeric, client's age.
*   **`job`**: Categorical, type of job (e.g., 'admin.', 'blue-collar', 'management').
*   **`marital`**: Categorical, marital status (e.g., 'married', 'single', 'divorced').
*   **`education`**: Categorical, education level (e.g., 'secondary', 'tertiary', 'primary').
*   **`default`**: Binary, indicates if the client has credit in default ('yes'/'no').
*   **`balance`**: Numeric, average yearly balance in euros.
*   **`housing`**: Binary, indicates if the client has a housing loan ('yes'/'no').
*   **`loan`**: Binary, indicates if the client has a personal loan ('yes'/'no').
*   **`contact`**: Categorical, contact communication type (e.g., 'cellular', 'telephone', 'unknown').
*   **`day`**: Numeric, last contact day of the month.
*   **`month`**: Categorical, last contact month of year (e.g., 'jan', 'feb', ..., 'dec').
*   **`duration`**: Numeric, last contact duration in seconds. **(Important Note: This attribute highly influences the output target 'y'. If duration=0, then 'y'='no'. You might want to discard this attribute if the goal is to build a realistic predictive model, as the duration is not known before the call is performed.)**
*   **`campaign`**: Numeric, number of contacts performed during this campaign for the client.
*   **`pdays`**: Numeric, number of days that passed after the client was last contacted from a previous campaign (-1 means not previously contacted).
*   **`previous`**: Numeric, number of contacts performed before this campaign for the client.
*   **`poutcome`**: Categorical, outcome of the previous marketing campaign (e.g., 'failure', 'success', 'other', 'unknown').

### **Target Variable:**
*   **`y`**: Binary, indicates if the client subscribed to a term deposit ('yes'/'no').
